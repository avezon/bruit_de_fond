```python
def set_detectors_channels(df, ch_RED = [1,2,3], ch_BLUE = [5,6,7], ch_YELLOW = [9,10,11], ch_WHITE = [12,13,14]):
    a=ch_RED[0]
    b=ch_RED[1]
    c=ch_RED[2]
    df_RED = df.query("sub_channel == @ch_RED[0] | sub_channel == @b | sub_channel == @c")
    a=ch_BLUE[0]
    b=ch_BLUE[1]
    c=ch_BLUE[2]
    df_BLUE = df.query("sub_channel == @a | sub_channel == @b | sub_channel == @c")
    a=ch_YELLOW[0]
    b=ch_YELLOW[1]
    c=ch_YELLOW[2]
    df_YELLOW = df.query("sub_channel == @a | sub_channel == @b | sub_channel == @c")
    a=ch_WHITE[0]
    b=ch_WHITE[1]
    c=ch_WHITE[2]
    df_WHITE = df.query("sub_channel == @a | sub_channel == @b | sub_channel == @c")
    
    return(df_RED,df_BLUE,df_YELLOW,df_WHITE)
```

```python
df_RED,df_BLUE,df_YELLOW,df_WHITE = set_detectors_channels(df, ch_RED = [1,2,3], ch_BLUE = [5,6,7], ch_YELLOW = [9,10,11], ch_WHITE = [12,13,14])
```
